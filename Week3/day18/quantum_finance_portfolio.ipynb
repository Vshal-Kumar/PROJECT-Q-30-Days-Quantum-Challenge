```python
from qiskit import QuantumCircuit
from qiskit.primitives import StatevectorSampler

# ----------------------------
# Step 1: Create Quantum Circuit
# 4 qubits represent 4 assets
# ----------------------------
qc = QuantumCircuit(4, 4)

# ----------------------------
# Step 2: Create Superposition
# Each qubit is in |0> and |1>
# representing both investment choices
# ----------------------------
qc.h(range(4))

# ----------------------------
# Step 3: Parameterized Rotations
# These angles determine the probability
# of selecting each asset
# ----------------------------
qc.ry(0.8, 0)   # Asset A
qc.ry(1.2, 1)   # Asset B
qc.ry(0.5, 2)   # Asset C
qc.ry(1.0, 3)   # Asset D

# ----------------------------
# Step 4: Measure All Qubits
# ----------------------------
qc.measure(range(4), range(4))

# Display the quantum circuit
print("Quantum Circuit:")
print(qc.draw())

# ----------------------------
# Step 5: Execute the Circuit
# ----------------------------
sampler = StatevectorSampler()

job = sampler.run([qc], shots=1024)
result = job.result()

# ----------------------------
# Step 6: Get Measurement Counts
# (Qiskit 2.4.2 uses register name 'c')
# ----------------------------
counts = result[0].data.c.get_counts()

print("\nPortfolio Measurement Counts:")
print(counts)

# ----------------------------
# Step 7: Display Portfolios
# ----------------------------
print("\nPossible Investment Portfolios:")

for portfolio, frequency in sorted(counts.items()):
    print(f"Portfolio {portfolio} : {frequency} times")
```
