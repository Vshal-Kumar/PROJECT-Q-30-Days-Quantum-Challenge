```python
from qiskit import QuantumCircuit
from qiskit_ibm_runtime import QiskitRuntimeService, Sampler

# 1. Access cloud hardware accounts
service = QiskitRuntimeService()

# 2. Pick the optimal, least busy machine available
backend = service.least_busy(simulator=False)

# 3. Define a standard Bell State circuit
qc = QuantumCircuit(2)
qc.h(0)
qc.cx(0,1)
qc.measure_all()

# 4. Broadcast the circuit instructions to IBM Cloud
sampler = Sampler(mode=backend)
job = sampler.run([qc])
result = job.result()
```
